Train ve Test olarak bir ayrım yapacağız. Bunu `sklearn.model_selection`'dan train_test_split() ile de yapabiliriz.

```python
train_split = int(len(X) * 0.8)
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]
```

```python
X_train
```

- `X_train, y_train = X[:N], y[:N]` : En baştan (0. indeksten) başla, N. indekse kadar git, ama N. indeksi DAHİL ETME.
- `X_test, y_test = X[N:], y[N:]` : N. indeksten başla (N. indeks DAHİL) ve verinin en sonuna kadar git.

```python
plt.scatter(X_train, y_train, c = "r", s = 4, label = "Training Data")
plt.scatter(X_test, y_test, c = "orange", s = 4, label = "Testing Data")
plt.show()
```

Şimdi ilk derin ağımı manuel bir şekilde yazacağım.

```python
from torch import nn

class SimpleLinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.weights = nn.Parameter(
            torch.randn(1, dtype = torch.float),    # Weights için rastgele değer veriyor.
            requires_grad = True     # Öğrenilebilir bir parametre olduğunu gösterir.
        )
        self.bias = nn.Parameter(
            torch.randn(1, dtype = torch.float),
            requires_grad = True
        )

    def forward(self, x : torch.Tensor) -> torch.Tensor:
        return self.weights * x + self.bias
```

```python
torch.manual_seed(42)

model_0 = SimpleLinearRegressionModel()
```

```python
list(model_0.parameters())
```

```python
model_0.state_dict()
```

Şimdi henüz sadece rastgele ağırlık ve bias değerine sahip olan modelimiz ile test setimiz üzerinden bir tahmin yapalım

```python
with torch.inference_mode():
    y_preds = model_0(X_test)
```

```python
len(X_test), len(y_preds)
```

```python
plt.scatter(X_train, y_train, c = "r", s = 4, label = "Training Data")
plt.scatter(X_test, y_test, c = "orange", s = 4, label = "Testing Data")
plt.scatter(X_test, y_preds, c = "b", s = 4, label = "Predictions")
plt.show()
```

Şimdi de Loss Fonksiyonumu ve Optimizer seçimimi yapacağım.

```python
loss_fn = nn.MSELoss()    # L1Loss -> MAE

optimizer = torch.optim.SGD(
    params = model_0.parameters(),
    lr = 0.001
)
```

```python
torch.manual_seed(42)

epochs = 120
train_loss_values = []
test_loss_values = []
epoch_count = []

for epoch in range(epochs):

    model_0.train()    # Train mode

    y_pred = model_0(X_train)    # 1. Forward Propagation

    loss = loss_fn(y_pred, y_train)    # 2. Calculate the loss

    optimizer.zero_grad()   # 3. Önceki gradyanı sıfırlar.

    loss.backward()    # 4. Back Propagation

    optimizer.step()   # 5. Ağırlıkları optimize eder.


    model_0.eval()   # Modeli test moduna alır

    with torch.inference_mode():

        test_pred = model_0(X_test)    # 1. Make Predictions

        test_loss = loss_fn(test_pred, y_test.type(torch.float))    # 2. Calculate the test loss

        if epoch % 5 == 0:
            epoch_count.append(epoch)
            train_loss_values.append(loss.detach().numpy())
            test_loss_values.append(test_loss.detach().numpy())
            print(f"Epoch : {epoch} | Train Loss: {loss} | Test Loss: {test_loss} ")
    
    
    
```

```python
plt.plot(epoch_count, train_loss_values)
plt.plot(epoch_count, test_loss_values)
plt.title("Training and test loss curves")
plt.ylabel("Loss")
plt.xlabel("Epochs")
plt.show()
```
