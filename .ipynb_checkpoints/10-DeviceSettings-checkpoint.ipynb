Bu MAC kullananlar içindir.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
device
```

```python
tensor = torch.tensor([1,2,3])
```

```python
tensor
```

```python
tensor.device
```

Bunu GPU yapmanın 3 yolu var.

### 1- Manuel

```python
tensor_on_gpu = tensor.to(device)
```

```python
tensor_on_gpu
```

```python
tensor_on_gpu.numpy()
```

Numpy'a çevirirken hata çıkar.

```python
tensor_numpy = tensor_on_gpu.cpu().numpy()
```

```python
tensor_numpy
```

### 2- Context Manager

```python
with torch.device(device):
    tensor2 = torch.tensor([1,2,3])
    layer = torch.nn.Linear(20,30)
```

```python
tensor2
```

```python
layer.weight.device
```

### 3- Set Default Device

```python
torch.set_default_device(device)
```

```python
tensor3 = torch.tensor([1,2,3])
layer2 = torch.nn.Linear(20,30)
```

```python
tensor3.device
```

```python
layer2.weight.device
```
